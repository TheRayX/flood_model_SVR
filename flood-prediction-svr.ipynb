```python
df=df.set_index(['date'],drop=['date'])
```

```python
X= df.drop(columns=['flood_magnitude'],axis=1)
```

```python
y=df['flood_magnitude']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=101)
```

```python
from sklearn.svm import SVR 
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
scaled_X_train=scaler.fit_transform(X_train)
```

```python
scaled_X_test=scaler.transform(X_test)
```

```python
model= SVR()
```

```python
param_grid ={
    'C':[0.001,0.003,0.01,0.02,0.08,0.1,0.3,0.5,0.55],
    'kernel':['linera','rbf','poly'],
    'gamma':['scale','auto'],
    'degree':[2,3,4,5,6,7,8],
    'epsilon':[0,0.01,0.1,0.3,0.5,1]
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid=GridSearchCV(model,param_grid,verbose=1)
```

```python
grid.fit(scaled_X_train,y_train)
```

```python
grid.best_params_
```

```python
grid_preds=grid.predict(scaled_X_test)
```

```python
from sklearn.metrics import mean_squared_error,mean_absolute_error 
```

```python
mean_squared_error(y_test,grid_preds)
```
